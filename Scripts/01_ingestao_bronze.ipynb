```python
# CÉLULA 1 - Leitura do arquivo CSV

caminho_csv = "/Volumes/workspace/projeto_integrador/dados_sus/ccmec25a64.csv"

df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .option("sep", ",")
    .csv(caminho_csv)
)

display(df)
```

```python
# CÉLULA 2 - Verificação da estrutura da base

df.printSchema()

print("Quantidade de registros:", df.count())
print("Quantidade de colunas:", len(df.columns))
print("Colunas:", df.columns)

display(df.limit(10))
```

```python
# CÉLULA 3 - Criação da tabela Bronze em formato Delta

df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "workspace.projeto_integrador.bronze_exames_citopatologicos"
    )
```

```python
# CÉLULA 4 - Validação da tabela Bronze criada

df_bronze = spark.table(
    "workspace.projeto_integrador.bronze_exames_citopatologicos"
)

display(df_bronze.limit(10))
```
